```python
import fastf1
import pandas as pd
fastf1.Cache.enable_cache('D:\\Programming\\Python\\AI\\F1 Fantasy\\cache')
```

```python
session = fastf1.get_session(2026, 'Hungary', 'R')
session.load(telemetry=False)

laps = session.laps
```

```python
laps = session.laps.copy()

data = laps[
    [
        "Driver",
        "LapNumber",
        "LapTime",
        "Compound",
        "TyreLife",
        "Stint",
        "Position",
        "PitInTime",
        "PitOutTime",
        "TrackStatus",
        "IsAccurate"
    ]
].copy()

data["LapTimeSeconds"] = data["LapTime"].dt.total_seconds()

data["RaceProgress"] = (
    data["LapNumber"] / data["LapNumber"].max()
)

data = data.dropna(
    subset=[
        "LapTimeSeconds",
        "Compound",
        "TyreLife",
        "Stint"
    ]
)

data = data[
    data["PitInTime"].isna() &
    data["PitOutTime"].isna()
]

data = data[data["IsAccurate"] == True]

data
```

```python
nor = data[data["Driver"] == "NOR"]

nor.head(20)
```

```python
stint = nor[nor["Stint"] == nor["Stint"].min()].copy()

stint = stint.pick_quicklaps()

stint[
    [
        "LapNumber",
        "LapTimeSeconds",
        "Compound",
        "TyreLife",
        "RaceProgress"
    ]
]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(
    stint["TyreLife"],
    stint["LapTimeSeconds"]
)

plt.xlabel("Tyre Life")
plt.ylabel("Lap Time (seconds)")
plt.title("Lap Time vs Tyre Life")

plt.show()
```

```python
from sklearn.linear_model import LinearRegression

X = stint[["TyreLife"]]
y = stint["LapTimeSeconds"]

model = LinearRegression()
model.fit(X, y)

slope = model.coef_[0]
intercept = model.intercept_

print("Degradation:", slope)
print("Base pace:", intercept)
```

```python
r2 = model.score(X, y)

print("R²:", r2)
```

```python
results = []

for (driver, stint_number), group in data.groupby(["Driver", "Stint"]):

    # Need at least a few laps
    if len(group) < 5:
        continue

    X = group[["TyreLife"]]
    y = group["LapTimeSeconds"]

    model = LinearRegression()
    model.fit(X, y)

    results.append({
        "Driver": driver,
        "Stint": stint_number,
        "Compound": group["Compound"].iloc[0],
        "Laps": len(group),
        "BasePace": model.intercept_,
        "Degradation": model.coef_[0],
        "R2": model.score(X, y)
    })

degradation_df = pd.DataFrame(results)

degradation_df
```

```python
X = stint[
    [
        "TyreLife",
        "RaceProgress"
    ]
]

y = stint["LapTimeSeconds"]

model = LinearRegression()
model.fit(X, y)

print("Coefficients:")
print(model.coef_)

print("R²:", model.score(X, y))
```

```python
dutchSprint = fastf1.get_session(2026, 'dutch', 'S')
dutchSprint.load(telemetry=False)
```

```python
dutchSprintLaps = dutchSprint.laps.copy()
rusLaps = dutchSprintLaps[dutchSprintLaps['Driver'] == 'NOR'].copy()
rusLaps = rusLaps.pick_quicklaps().copy()
rusLaps["LapTimeSeconds"] = rusLaps["LapTime"].dt.total_seconds()

rusLaps = rusLaps[
    [
        'Driver',
        'LapTimeSeconds',
        'TyreLife'
    ]
]
rusLaps = rusLaps.dropna()
rusLaps
```

```python
import matplotlib.pyplot as plt
import numpy as np

m, b = np.polyfit(rusLaps["TyreLife"], rusLaps["LapTimeSeconds"], 1)

plt.figure(figsize=(8, 5))

plt.scatter(
    rusLaps["TyreLife"],
    rusLaps["LapTimeSeconds"]
)
plt.axline(xy1=(0, b), slope=m, color="red", linestyle="--", label=f"Best Fit: y={m:.2f}x+{b:.2f}")

plt.xlabel("Tyre Life")
plt.ylabel("Lap Time (seconds)")
plt.title("Lap Time vs Tyre Life")

plt.show()
```
