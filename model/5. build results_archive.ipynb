```python
import fastf1
import pandas as pd

# Enable FastF1 cache
fastf1.Cache.enable_cache('D:\\Programming\\Python\\AI\\F1 Fantasy\\cache')

results_archive = pd.read_csv("data/races/results_archive.csv")
```

```python
n_rounds = 24
year = 2022
#results_archive = pd.DataFrame()
for round in range(1, n_rounds+1):
    session = fastf1.get_session(year, round, 'r')
    session.load(laps=False, telemetry=False, weather=False, messages=False)
    results = session.results.copy()
    results['Year'] = year
    results['Round'] = round
    results['Track'] = session.event['Location']
    results_archive = pd.concat([results_archive, results], ignore_index=True)
```

```python
def is_dnf(status):
    finished_like = {'Finished'}
    if status in finished_like:
        return False
    if status.startswith('+') or 'Lap' in status:   # e.g. "+1 Lap" — still classified as finished
        return False
    return True   # Accident, Retired, Engine, Collision, Disqualified, etc.

results_archive['dnf'] = results_archive['Status'].apply(is_dnf)
results_archive.to_csv("data/races/results_archive.csv", index=False)
```
